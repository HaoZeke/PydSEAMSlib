# Classify ice with d-SEAMS in a notebook

Install the wheel, download one frame of the engine's cubic mW lattice, and read the seeded cage assignment, the CHILL+ labels and a topology fingerprint from Python. Runs on Colab with the wheel from PyPI.

```python
%pip install -q pydseamslib ase
```

```python
import urllib.request, pathlib
url = 'https://raw.githubusercontent.com/d-SEAMS/seams-core/main/input/traj/mW_cubic.lammpstrj'
path = pathlib.Path('mW_cubic.lammpstrj')
if not path.exists():
    urllib.request.urlretrieve(url, path)
print(path.stat().st_size, 'bytes')
```

```python
from pydseams import Frame
frame = Frame.from_file(str(path), atom_type=1)
print(frame.cloud.nop, 'molecules')
score = frame.seeded_affiliation(ring_adjacent=True)
print('hexagonal-cage molecules', sum(score.hc), 'double-diamond', sum(score.ddc))
```

```python
counts = frame.chill_plus()
print(counts)
```

```python
fp = frame.fingerprint(hops=3)
print(fp.method, 'classes:', len(fp.classes), 'ring census:', list(fp.ringCensus))
lib = frame.topology_library('Ic', hops=3)
print(frame.classify_topology(lib, hops=3).counts)
```

Every molecule of the cubic lattice falls in one topology class and the library built from the frame names all of them. Point `Frame.from_file` at your own LAMMPS dump, or build a frame from ASE with `pydseams.from_ase(atoms, select=('O',))`, and the same calls apply. The [features how-to](https://d-seams.github.io/PydSEAMSlib/howto/features.html) turns a trajectory into per-frame vectors for PyEMMA and deeptime, and `pydseams.adapters` carries an MDAnalysis analysis and an OVITO modifier.
